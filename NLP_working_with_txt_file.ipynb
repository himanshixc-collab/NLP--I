```python
#adding book names in write mode
file = open("library.txt","w")

file.write("Panchtantra\n")
file.write("Diary of a Wimpy kid\n")
file.write("Geronimo stilton\n")
file.write("Harry Potter\n")

file.close()
print("written successfully")
```

```python
#reading the entire txt file
with open('library.txt','r')as file:
    data=file.read()
    print(data)
```

```python
#Appending data to file
with open ("library.txt","a") as file:
  file.write("Dog Man\n")
  file.write("Goosebumps\n")
print("Books added")
```

```python
#writing book details in the file

sequence_number= input("Enter the sequence number: ")
serie_name= input("Enter the name of the series: ")
with open("books.txt","w") as file:
  file.write("Name: " + sequence_number+ "\n")
  file.write("Age: " +serie_name + "\n")

print("Information saved")
```

```python
#counting the number of files

with open ("library.txt", "r") as file:
  lines= file.readlines()

print ("Number of books:", len(lines))
```

```python
#counting words in a text file

with open ("library.txt", "r") as file:
  data= file.read()
words = data.split()

print("Number of words:", len(words))
```

```python
#searching word in the file
search_word = input("enter word to search:")

with open("library.txt","r") as file:
  data = file.read()

if search_word.lower() in data.lower():
  print ("word found.")
else:
  print("word not found")
```

```python
# counting occurence of a word

search_word = input ("enter word: ")
with open ("library.txt","r") as file:
    data = file.read()

count = data.lower().count(search_word.lower())

print("occurences: ",count)
```

```python
#copying contents from 1 file to another

with open ("library.txt","r") as source:
  data = source.read()

with open("backup.txt","w") as destination:
  destination.write(data)

print("file copied successfully.")
```
