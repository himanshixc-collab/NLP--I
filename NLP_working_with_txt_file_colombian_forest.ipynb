```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
#tokenization with token index
doc = nlp(data)

for token in doc:
    print(token.i, token.text)
```

```python
from nltk.stem import PorterStemmer
stemmer= PorterStemmer()
```

```python
#stemming (removing punctuations before stemming)

doc= nlp(data)

for token in doc:
  if token.is_alpha:
    print(token.text,"->", stemmer.stem(token.text))
```

```python
#lemmatization with stopword removal

doc= nlp (data)
words=[]

for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.lemma_)

print(words)
```

```python
#extract only stopwords
doc= nlp(data)
stopwords=[]

for token in doc:
  if token.is_stop:
    stopwords.append(token.text)

print(stopwords)
```

```python
#POS Tagging (extract proper nouns)
doc= nlp(data)

for token in doc:
  if token.pos_ == "PROPN":
    print(token.text)
```

```python
#NER (locations)
doc = nlp(data)
for ent in doc.ents:
    print(ent.text, ent.label_)
```
